```python
plt.figure(figsize=(10, 5))

state_count = df['state'].value_counts().head(10)

sns.barplot(
    x=state_count.values,
    y=state_count.index
)

plt.title('Top 10 States by Customer Count')
plt.xlabel('Number of Customers')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```

```python
gender_country = pd.crosstab(
    df['country'],
    df['gender']
)

gender_country.plot(
    kind='bar',
    stacked=True,
    figsize=(9, 5)
)

plt.title('Gender-wise Customer Distribution by Country')
plt.xlabel('Country')
plt.ylabel('Customer Count')
plt.legend(title='Gender')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
completeness = (df.notnull().mean() * 100).sort_values()

plt.figure(figsize=(9, 5))
sns.barplot(x=completeness.values, y=completeness.index)
plt.title('Customer Data Completeness')
plt.xlabel('Completeness (%)')
plt.ylabel('Columns')
plt.xlim(0, 100)

for i, value in enumerate(completeness.values):
    plt.text(value + 1, i, f'{value:.1f}%')

plt.tight_layout()
plt.show()
```

```python
# Create Status column in df2

df2["Status"] = df2["cancellation_date"].apply(
    lambda x: "Churned" if pd.notna(x) else "Retained"
)

print(df2.columns)
```

```python
print(df2[["customerid", "cancellation_date", "Status"]].head())
```

```python
df2["Status"].value_counts().plot(
    kind="pie", autopct="%1.1f%%",
    startangle=90, wedgeprops={"width": 0.4}
)
plt.title("Overall Customer Churn Rate")
plt.ylabel("")
plt.show()
```

```python
pd.crosstab(df2["plan_type"], df2["Status"]).plot(
    kind="bar", stacked=True, figsize=(9,5)
)
plt.title("Churn by Subscription Plan")
plt.xlabel("Plan Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.show()
```

```python
df2["cancellation_reason"].dropna().value_counts().head(8).plot(
    kind="barh", figsize=(9,5)
)
plt.title("Top Reasons for Customer Churn")
plt.xlabel("Number of Customers")
plt.ylabel("Cancellation Reason")
plt.show()
```
